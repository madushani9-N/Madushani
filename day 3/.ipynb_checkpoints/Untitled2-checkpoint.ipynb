# Boolean Indexing and Slicing

```python
random_integers = np.random.randint(1, 101, 20) #randint kiynne intiger numbers one nisa dana eka
random_integers
```

```python
random_integers % 2   # % meka dala even or odd hoyanna


# 1 kiyanne odd 0 kiyanne even
```

```python
even_numbers =random_integers[random_integers % 2 == 0]     #even numbers withrak ganna
even_numbers        
```

```python
sliced_array = even_numbers[2:10]    # 2 weni thana idan 19 weni thana wena kam even numbers one nam
sliced_array 
```

# Vectorized Operations

```python
arr1 = np.random.randint(0,10,4)
arr2 = np.random.randint(0,10,4)

print(f"the two arrays are:\n{arr1}\n{arr2}")
                   
```

```python
print(f"the sum of two arrays is {arr1+arr2}")
print(f"the difference of two arrays is {arr1-arr2}")
print(f"the multiplication of two arrays is {arr1*arr2}")
```

```python
np.dot(arr1,arr2)    # ekathuwa wage
```

```python
(arr1*arr2).sum()
```
