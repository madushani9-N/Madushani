# Bar Chart 


```python


categories = ['company B', 'company B', 'company C','company D']  #use lists
values = [20, 15, 10, 40]
colors = ["green","blue","red","orange"]




plt.bar(categories, values,color=colors) #color eka danakota wenama color = krala danna one mokada eka parameter eken eha nisa 
plt.title('Bar Chart')
plt.xlabel('Categories')
plt.ylabel('revenues')
plt.xticks(rotation=45) # Rotate for better readability

plt.show()
```
