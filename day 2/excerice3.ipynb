```python
def is_even(a):
    """
    this function returns true if the number is even,otherwise it returns false
    """
    result = a%2
    if result ==1:
        return False
    else:
        return True


is_even(10)
```

```python
is_even(11)
```

```python
def is_even_2(a):
    """
version 2 
    """
   
    if a%2 ==1:
        return False
    else:
        return True


is_even(10)
```

```python
def is_even_3(a):
    """
version 3
    """
   
    if a%2 ==0:
        return False
    
    return True

is_even(10)
```

```python
def is_even_4(a):
    """
version 4
    """
   
    return (a%2 ==0) #return true if the result is 0,otherwise return false
       

is_even(10)
```
